```python
import pandas as pd
import numpy as np
import os
import sys
import json
```

```python
# read preferences.json
with open('preferences_clean.json') as f:
    allpreferences = json.load(f)
users = allpreferences['user']
course_sem_ids = allpreferences['course_sem_ids']
preferences = allpreferences['preferences']
```

```python
print("Users: ", users)
print("Course Sem IDs: ", course_sem_ids)
print("Preferences: ", preferences)
```

```python
# dict for users: id, name, email, num_req_courses
user_dict = {}
for user in users.keys():
    user_dict[int(user)] = {
        'name': users[user],
        'email': users[user] + '@arizona',
        'num_req_courses': 1
    }
user_dict
```

```python
# course_sem
#     id SERIAL PRIMARY KEY,
#    course_id INT NOT NULL,
#   semester VARCHAR NOT NULL,
course_sem_dict = {}
for course_sem in course_sem_ids:
    course_sem_dict[course_sem] = {
        'course_id': course_sem,
        'semester': 'Fall 2024'
    }
course_sem_dict
```

```python
# -- Preference Table
# CREATE TABLE IF NOT EXISTS match_schema.preferences (
#     id SERIAL PRIMARY KEY,
#     semester VARCHAR NOT NULL,
#     user_id INT NOT NULL,
#     course_sem_id INT NOT NULL ,
#     preference_level VARCHAR NOT NULL,
#     preference_weight INT NOT NULL
# );
preference_dict = {}
for preference in preferences:
    user_id = preference['user_id']
    course_sem_id = preference['course_sem_id']
    preference_level = ''
    if preference['level'] == 'g':
        preference_level = 'green'
    elif preference['level'] == 'y':
        preference_level = 'yellow'
    elif preference['level'] == 'r':
        preference_level = 'red'
    preference_weight = 1
    preference_dict[preference['id']] = {
        'semester': 'Fall 2024',
        'user_id': user_id,
        'course_sem_id': course_sem_id,
        'preference_level': preference_level,
        'preference_weight': preference_weight
    }
preference_dict
```

```python
# -- Course Table
# CREATE TABLE IF NOT EXISTS match_schema.courses (
#     id SERIAL PRIMARY KEY,
#     name VARCHAR NOT NULL,
#     number VARCHAR NOT NULL,
#     campus VARCHAR NOT NULL
#     -- Add other course-specific attributes here
# );
with open('new_preferences.json') as f:
    courses_json = json.load(f)
courses_json
courses_dict = {}
for course_id in courses_json.keys():
    cid = int(course_id)
    courses_dict[cid] = {
        'name': courses_json[course_id],
        'number': course_id,
        'campus': 'Main'
    }
courses_dict
```

```python
import psycopg2

# Establish a connection to the PostgreSQL database
conn = psycopg2.connect(
    dbname="match_db",
    user="match_user",
    password="swifty",
    host="localhost",
    port="5432"
)
cursor = conn.cursor()

# Insert data into the users table
for user_id, user_data in user_dict.items():
    cursor.execute("""
        INSERT INTO match_schema.users (id, name, email, num_req_courses)
        VALUES (%s, %s, %s, %s)
        ON CONFLICT (id) DO NOTHING;
    """, (user_id, user_data['name'], user_data['email'], user_data['num_req_courses']))

# Insert data into the course_sem table
for course_sem_id, course_sem_data in course_sem_dict.items():
    cursor.execute("""
        INSERT INTO match_schema.course_sem (id, course_id, semester)
        VALUES (%s, %s, %s)
        ON CONFLICT (id) DO NOTHING;
    """, (course_sem_id, course_sem_data['course_id'], course_sem_data['semester']))

# Insert data into the courses table
for course_id, course_data in courses_dict.items():
    cursor.execute("""
        INSERT INTO match_schema.courses (id, name, number, campus)
        VALUES (%s, %s, %s, %s)
        ON CONFLICT (id) DO NOTHING;
    """, (course_id, course_data['name'], course_data['number'], course_data['campus']))

# Insert data into the preferences table
for preference_id, preference_data in preference_dict.items():
    cursor.execute("""
        INSERT INTO match_schema.preferences (id, semester, user_id, course_sem_id, preference_level, preference_weight)
        VALUES (%s, %s, %s, %s, %s, %s)
        ON CONFLICT (id) DO NOTHING;
    """, (preference_id, preference_data['semester'], preference_data['user_id'],
          preference_data['course_sem_id'], preference_data['preference_level'], preference_data['preference_weight']))

# Commit the transaction and close the connection
conn.commit()
cursor.close()
conn.close()
```
